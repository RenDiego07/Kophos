```python
import zipfile
import pandas as pd
from collections import Counter
import os

zip_path = "/content/drive/MyDrive/ASL_Citizen.zip"
csv_salida = "/content/conteo_palabras_asl_citizen.csv"

print("Analizando el archivo ZIP en memoria (sin extraer videos)...")

try:
    with zipfile.ZipFile(zip_path, 'r') as z:
        archivos = z.namelist()

        # 1. Filtrar solo los archivos de video
        videos = [f for f in archivos if f.endswith('.mp4')]
        print(f"Total de videos encontrados: {len(videos)}")

        # 2. Extraer la palabra de cada nombre de archivo
        palabras = []
        for v in videos:
            # Aislar el nombre del archivo (ej: 5086723486875651-RUN.mp4)
            nombre_archivo = os.path.basename(v)

            # Cortar por el guion '-' y quitar la extensión '.mp4'
            if '-' in nombre_archivo:
                palabra = nombre_archivo.split('-')[-1].replace('.mp4', '')
                palabras.append(palabra)

        # 3. Contar cuántos videos hay por cada palabra
        print("Contando frecuencias...")
        conteo = Counter(palabras)

        # 4. Convertir a DataFrame de Pandas y ordenar
        df_conteo = pd.DataFrame(conteo.items(), columns=['Palabra', 'Cantidad_Videos'])
        df_conteo = df_conteo.sort_values(by='Cantidad_Videos', ascending=False)

        # 5. Exportar a CSV
        df_conteo.to_csv(csv_salida, index=False)

        print(f"\n¡Análisis completado! Se encontraron {len(df_conteo)} palabras únicas.")
        print("\n--- Top 5 palabras con más videos ---")
        print(df_conteo.head(5).to_string(index=False))
        print(f"\n✅ El archivo CSV se ha guardado en: {csv_salida}")
        print("Ya puedes descargarlo desde el panel izquierdo de Colab y abrirlo en tu Excel/Numbers.")

except FileNotFoundError:
    print("Error: No se encontró el archivo ZIP. Verifica la ruta.")
```

```python
import zipfile
import os
import shutil
from collections import defaultdict

# 1. Definir rutas y objetivo
zip_path = "/content/drive/MyDrive/ASL_Citizen.zip"
extract_path = "/content/Mis_30_Clases"
zip_descarga = "/content/Mis_30_Clases_Descarga"

# Tu léxico final depurado, eliminando CHAT, UNDERSTAND, TO, WRITE y ASK
palabras_objetivo = {
    "BAD", "BECOME", "FIND", "FINISH", "GOOD",
    "HELLO", "HELP", "HOSPITAL", "KNOW", "LAUGH",
    "LIKE", "MAKE", "ME", "MORE", "NEED",
    "NO", "PLAY", "PLEASE", "READ", "SHOW",
    "START", "TELL", "THINK", "WANT", "WHERE",
    "WHO", "WHY", "YES", "YOU"
}

os.makedirs(extract_path, exist_ok=True)
conteo_extraido = defaultdict(int)

print(f"Buscando {len(palabras_objetivo)} clases en la memoria. Esto tomará unos segundos...")

try:
    with zipfile.ZipFile(zip_path, 'r') as z:
        archivos = z.namelist()

        # Filtrar los videos que nos interesan
        videos_a_extraer = []
        for f in archivos:
            if f.endswith('.mp4'):
                # Extraer la palabra del formato "123456789-PALABRA.mp4"
                nombre_base = os.path.basename(f)
                if '-' in nombre_base:
                    # Asegurar mayúsculas para evitar problemas de coincidencia
                    palabra = nombre_base.split('-')[-1].replace('.mp4', '').upper()
                    if palabra in palabras_objetivo:
                        videos_a_extraer.append((f, palabra))

        print(f"Se encontraron {len(videos_a_extraer)} videos en total que coinciden con tus clases.")
        print("Iniciando extracción al disco de Colab...")

        # Extraer y contar
        for ruta_original, palabra in videos_a_extraer:
            z.extract(ruta_original, extract_path)
            conteo_extraido[palabra] += 1

    print("\n--- Resumen de Extracción ---")
    for palabra, cantidad in sorted(conteo_extraido.items(), key=lambda x: x[1], reverse=True):
        print(f"{palabra}: {cantidad} videos")

    print("\nComprimiendo la carpeta resultante para descarga...")
    shutil.make_archive(zip_descarga, 'zip', extract_path)

    print(f"\n✅ ¡Éxito! El archivo listo para descargar a tu Mac es: {zip_descarga}.zip")

except FileNotFoundError:
    print("Error: No se encontró el archivo ZIP. Verifica que tu Drive esté montado correctamente.")
```
